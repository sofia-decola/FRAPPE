```python
import stringdb as sdb
import pandas as pd
```

# Example Usage/Tutorial

Datasets acquired through [ECHO](https://pmc.ncbi.nlm.nih.gov/articles/PMC7523678/) preprocessing have a format that StringDB is designed to take into account specifically, but any pandas DataFrame with expression data can work 

[Github Documentation Here ](https://github.com/delosh653/ECHO)

An example of this formatting is the following:

| Protein |	 Timepoint 1| 	Timepoint 2	| Timepoint 3| 	Timepoint 4| 	Timepoint 5| Timepoint 6 |
| ------------- |-------------|-------------|-------------|-------------|-------------|-------------|
| Sample 1 |	1.633117905| 1.4448585	| 	1.513810213| 	1.309553546 | 	1.302488129| 1.8384729 	|	
| Sample 2 | 	-0.630319173| -0.4228347	| 	-0.510500938| -0.5228194 	| 	-0.543457041| 	-0.448383157|		
| Sample 3	| -0.780221402| 0.17483084	| 0.238884	| 	0.178429468| 	0.306513019| 	1.376226634|



Necessary clustering is done first and applied to filtered data, StringDB **CANNOT** be run without these precursors

```python
df = pd.read_csv('File_path_to_data_here.csv')
df_formatted = sdb.echo_to_format(df) #IF ECHO format, if not, can be replaced with format above
clust_labels = sdb.clustering_rawdata(df_formatted, ids = list(df_formatted["Gene Name"])) #column of dataframe with protein names, ECHO uses 'Gene Name'
```

String DB Analysis itself for an individual protein

```python
sa = sdb.stringdb_analysis(cl_df = clust_labels, protein = "INSERT_GENE_NAME") #pandas DataFrame output
#defaults applied: savefile = 'stringdb.csv', organism = 'mus muculus', databases = 'all'
```

Or a cluster of proteins (from the clustering_rawdata function)

i.e. say you want to analyze all proteins from the same cluster (cluster number 4 in example below)

```python
sa = sdb.stringdb_analysis(cl_df = clust_labels, protein = 4) #pandas DataFrame output
#defaults applied: savefile = 'stringdb.csv', organism = 'mus muculus', databases = 'all'
```

# Altering StringDB Parameters

### Example call using a specific database

Specifying database parameter to one of the following options:
['COMPARTMENTS', 'Component',
        'Function', 'InterPro', 'KEGG', 'Keyword', 'MPO', 'NetworkNeighborAL', 'PMID', 'Pfam', 'Process', 'RCTM', 'SMART', 'TISSUES', 'WikiPathways']
  

```python
sa = sdb.stringdb_analysis(cl_df = clust_labels, protein = "INSERT_GENE_NAME", database = 'KEGG') #pandas DataFrame output
#defaults applied: savefile = 'stringdb.csv', organism = 'mus muculus'
```

### Example call using a specified organism

Predefined options (strings) include: 'mus musculus',
        'homo sapiens',
        'drosophila melanogaster',
        'neurospora crassa',
        'danio rerio',
        'arabidopsis thaliana'

```python
#using a predefined option
sa = sdb.stringdb_analysis(cl_df = clust_labels, protein = "GENE_NAME", savefile = 'Saved_File_Output.csv', organism = 'homo sapiens', database = 'all') #pandas DataFrame output
sa.head() #pandas DataFrame output
```

OR any input NCBI Taxon ID (int) for a species included on the STRING database

```python
#inputting an NCBI Taxon ID, for example 511145 indicates E. Coli strain K-12
sa = sdb.stringdb_analysis(cl_df = clust_labels, protein = "GENE_NAME", savefile = 'Saved_File_Output.csv', organism = 511145, database = 'all') #pandas DataFrame output
sa.head() #pandas DataFrame output
```
