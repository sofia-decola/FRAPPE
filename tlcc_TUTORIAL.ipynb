# Time Lagged Cross Correlation (TLCC)


Proteomics Data Analysis Package FRAPPE: Framework for Rhythmic Analysis of Preliminary Proteomics Experiments
Tools to analyze Time Lagged Cross Correlation (TLCC) using pandas & numpy

Time Lagged Cross Correlation: A set of time series data analysis tools that compares correlation at different time lags for two datasets of proteomics time series data,
with the two data sets representing proteomics data in different experimental groups

Input Data Set Required Setup:
- any .csv file of data that includes a column of protein names as well as columns for each time point.
- the protein_col name must be consistent across datasets (default label is 'Gene Name')

This collection of functions can run TLCC on individual proteins or whole datasets. The goal is to analyze the optimal time lag distribution across two datasets,
one from a control condition and one from an experimental condition


```python
import tlcc 
```

# Example Python Usage/Tutorial

Running TLCC on a dataset

```python
csv1 = "Control_Group_Data.csv"
csv2 = "Experimental_Group_Data.csv"
#for data cleaning purposes, names of every column in the data that you want to EXCLUDE from TLCC
#i.e. any other data that is not the protein name or time-series data
#this can be left to default value if the dataframe is being inputted as pre-cleaned
cols= ["Convergence","Iterations","Amplitude.Change.Coefficient","Oscillation Type","Initial.Amplitude","Radian.Frequency","Period","Phase Shift","Hours Shifted","Equilibrium Value","Slope","Tau","P-Value","BH Adj P-Value","BY Adj P-Value","Original TP16.1","Original TP16.2","Original TP16.3","Original TP19.1","Original TP19.2","Original TP19.3","Original TP22.1","Original TP22.2","Original TP22.3","Original TP25.1","Original TP25.2","Original TP25.3","Original TP28.1","Original TP28.2","Original TP28.3","Original TP31.1","Original TP31.2","Original TP31.3","Original TP34.1","Original TP34.2","Original TP34.3","Original TP37.1","Original TP37.2","Original TP37.3","Original TP40.1","Original TP40.2","Original TP40.3"]
#columns above are standard ECHO columns except for "Fitted Time Point" columns
output_file = "Name_Of_Output_File" #does not need .csv ending
lag = 3 #how far apart the time points are in the dataset
corr= 0 #Spearman correlation)
tlcc.tlcc_datasets(csv1, csv2, output_file, cols, corr, lag, protein_col='Gene Name') #runs TLCC on the common proteins in datasets csv1 and csv2
```

### Plotting a TLCC file

```python
tlcc.plot_tlcc_histogram("Name_Of_Output_File.csv", "Control Condition String", "Experimental Condition String", lag) #Simple histogram output of TLCC results
# condition strings are for plot title
```
